```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path="/Users/DELL/Downloads/Social_Media_Engagement.csv"
```

```python
df=pd.read_csv(file_path)
```

```python
df.shape
```

```python
df.info()
```

```python
df.duplicated().value_counts()
```

```python
print(df.isnull().sum())
```

```python
platform_counts=df['Platform'].value_counts()
print("Platform_counts:\n",platform_counts)
```

```python
sns.countplot(x='Platform',data=df,palette='pastel')
plt.title('Platform_count')
plt.xlabel('Platform')
plt.ylabel('Platform_count')
plt.show()
```

```python
sns.lineplot(x='Platform', y='Posts Per Week', data=df, estimator='mean', marker='o')

plt.title("Average Posts Per Week by Platform")
plt.xlabel("Platform")
plt.ylabel("Posts Per Week")

plt.show()
```

```python
sns.barplot(x='Platform', y='Campaign Reach', data=df, estimator=sum)
plt.title("Campaign reach by platform")
plt.show()
```

```python
sns.barplot(x='Platform', y='Engagement Rate', data=df, estimator='mean')
plt.title("Engagement Rate by Platform")
plt.show()
```

```python
sns.barplot(x='Platform', y='Posts Per Week', data=df, estimator='mean')

plt.title("Average Posts Per Week by Platform")
plt.show()
```

```python
df['Reach_scaled'] = df['Campaign Reach'] / df['Campaign Reach'].max()

plt.figure()

sns.histplot(df['Engagement Rate'], bins=10, kde=True, label='Engagement Rate')
sns.histplot(df['Reach_scaled'], bins=10, kde=True, label='Reach (scaled)')

plt.legend()
plt.title("Distribution of Engagement Rate vs Campaign Reach")

plt.show()
```

```python
g = df.groupby('Platform')['Ad Spend (USD)'].sum()

plt.pie(g, labels=g.index, autopct='%1.1f%%')
plt.title("Ad Spend Distribution")
plt.show()
```

```python
sns.histplot(df['Engagement Rate'], kde=True)
plt.title("Engagement Rate Distribution")
plt.show()
```
