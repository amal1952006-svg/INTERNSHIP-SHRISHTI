```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,14,12,18,20]
plt.plot(x,y,marker='o')
plt.title("simple line chart")
plt.xlabel("time")
plt.ylabel("distance")
plt.grid(True)
plt.show()
```

```python
students=["A","B","C","D","E"]
marks=[85,70,90,65,95]
plt.bar(students,marks)
plt.title("student marks")
plt.xlabel("student")
plt.ylabel("marks")
plt.show()
```

```python
students=["A","B","C","D","E"]
marks=[85,70,90,65,95]
plt.bar(students,marks,color=["red","blue","green","orange","purple"])
plt.title("student marks")
plt.xlabel("students")
plt.ylabel("marks")
plt.show()
```

```python
activities=["study","exercise","sleep","hobbies"]
hours=[4,1,7,2]
plt.pie(hours,labels=activities,autopct="%1.1f%%")
plt.title("daily activity distribution")
plt.show()
```
